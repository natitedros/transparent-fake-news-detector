The dataset is then downloaded from Huggingface. We add trust_remote_code=True because there is a remote code that needs to be run automatically by our machine to access the dataset.

```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset("ucsbnlp/liar", trust_remote_code=True)
print(dataset)
```

```python
train_df = dataset['train'].to_pandas()
val_df = dataset['validation'].to_pandas()
test_df = dataset['test'].to_pandas()

print("Train shape:", train_df.shape)
print("\nSample rows:")
print(train_df[['label', 'statement', 'speaker', 'subject']].head(5))
print("\nLabel distribution (train):")
print(train_df['label'].value_counts().sort_index())
print("\nLabel dtype:", train_df['label'].dtype)
```
